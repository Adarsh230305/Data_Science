### Missing Data

### Finding Missing Data

```python
data={
    'A':[1,2,3,np.nan],
    'B':[1,np.nan,np.nan,2],
    'C':[1,np.nan,np.nan,np.nan],
    'D':[1,2,3,4]
}
df=pd.DataFrame(data)
print(df)
print('\n')
print(df.isna())
print('\n')
print(df.isna().sum()) # Count the no.of null values in each row
print(df.isna().any()) # tell whether atleat a null is present in row or not


```

### Removing Missing Data

```python
print(df)
print('\n')
print(df.dropna()) # it will just drop row which has nan,not remove from origanal data
print('\n')
df
```

```python
print(df.dropna(thresh=2)) # thresh =2 means row with 2 atleast non-null value
```

### Filling Missing Data
(when we write inplace=True they only
the orignal Data Frame gets,change both in fillna() ,dropna())

```python
# fillna() is used to fill data in column wise
print(df)

print(df.fillna(0)) # replace nan with zeros
print('\n')
values={
    'A':10,
    'B':20,
    'C':30,
    'D':40
}
print(df.fillna(value=values)) # fill null the particular values
print('\n')
print(df.fillna(df.mean())) # fill null with mean of that column
```

### Merging two dataframes
(there should be a comman thing in both df to merge)

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})
# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
print(employees)
print('\n')
print(salaries)
```

```python
pd.merge(employees,salaries,on="employee_id",how="inner") # inner merge on basis of employees ,no null value will be there
```

```python
pd.merge(employees,salaries,on="employee_id",how="outer") # outer merge on basis of employees ,null value will be there
```

```python
pd.merge(employees,salaries,on="employee_id",how="left") # left merge on basis of employees ,null value will be there
```

```python
pd.merge(employees,salaries,on="employee_id",how="right") # right merge on basis of salaries,null value will be there
```

### Group by

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
```

### Pivot table

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index='Product',columns='Region',aggfunc='median') # by default aggfunc is mean,but here we change it to mean
# note since we are using aggeration here,we need to provide numeric values in "values"
```

### Cross Tab

#### Extract no. of episodes from data and make a new columns of episodes

```python
df['Title']
```

```python
def episode_extract(txt):
  data=""
  check=False
  for i in txt:
    if i==')':
      check=False
      break
    if i=='(':
     check=True
     continue
    if check==True:
      data+=i
  return data
df['episodes']=df['Title'].apply(episode_extract)
df['episode']=df['episode'].str.replace(" eps","")
df['episode']=df['episode'].astype(int)
df
```

#### Make a new column for time stamps

```python
df['Title'][1]
def time_stamps(txt):
  data=""
  for i in range(len(txt)):
    if txt[i]==')':
      for j in range(i+1,i+20):
        data+=txt[j]
  return data
  # calculating month count
def count_months(date_range):
    start, end = [x.strip() for x in date_range.split('-')]

    start = pd.to_datetime(start, format='%b %Y')
    end = pd.to_datetime(end, format='%b %Y')

    # Inclusive month count
    return (end.year - start.year) * 12 + (end.month - start.month) + 1
df['time_stamps']=df['Title'].apply(time_stamps) # time stamps
df['months_count']=df['time_stamps'].apply(count_months) # month count
df
```
