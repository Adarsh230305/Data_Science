## Basic of Matplot

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,10)
y=x**2
```

```python
plt.plot(x,y)
```

### subplot

```python
plt.subplot(2,2,1) # 2,2,1 means canvas is divide into two rows two column ,1 represents the 1st row & column
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
#
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.subplot(1,2,2)
plt.plot(y,x)
```

### figure

```python
fig=plt.figure(figsize=(10,10))
ax1=fig.add_axes([0.1,0.1,0.8,0.8]) # [distance from left end,distance from bottum,width,height]
ax1.plot(x,y)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('main')

ax2=fig.add_axes([0.2,0.4,0.3,0.3])
ax2.plot(x,x)
ax2.set_xlabel('x')
ax2.set_ylabel('x')
ax2.set_title('sub')
```

### types of plots

```python
plt.scatter(x,y)
```

```python
from random import sample
data = sample(range(1,1000),100)
plt.hist(data)
```

```python
plt.boxplot(x)
```

### Saving Plot

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```

## Basic of Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### Types of Plot
1.distribution plots(used for quantitative items)

 a.distplot/histplot

 b.jointplot

 c.pairplot

 d.rugplot

2.catogory plots

#### distribution plots

##### distplot/histplot

```python
df=sns.load_dataset('tips')
df
```

```python
sns.distplot(df['total_bill']) # here kde is true by default ,KDE stands for Kernel Density Estimation, a statistical method used to visualize the continuous probability density distribution of a dataset.
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'])
plt.subplot(1,2,2)
sns.histplot(df['tip'])
```

##### joint plot

```python
sns.jointplot(x='total_bill',y='tip',data=df)
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind ='reg')
```

##### pair plot

```python
sns.pairplot(df,hue='sex',palette='rainbow')
```

```python
sns.pairplot(df,hue='size',palette='rainbow')
```

##### rug plot

```python
sns.rugplot(df['total_bill'])
```
