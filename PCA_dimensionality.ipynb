```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
```

```python
X, y = make_blobs(n_samples=500, centers=3, n_features=5,random_state=0)
X
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(X)
```

```python
PCA_reduction=PCA(n_components=2)
PCA_X=PCA_reduction.fit_transform(x_scaled)
PCA_X
```

```python
pca_df=pd.DataFrame(PCA_X,columns=['PCA_1','PCA_2'])
pca_df['label']=y
```

```python
sns.scatterplot(x=pca_df['PCA_1'],y=pca_df['PCA_2'],hue=pca_df['label'],palette='Set2')
```
